##Normalized Images

```python
X_train.shape,X_test.shape
```

```python
y_train.shape,y_test.shape
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
# normalizing is neccessary when we have images based data
```

```python
X_train
```

```python
# grayscale 0,1 (black,white)
# rgb (red,green,blue)
```

```python
X_train = X_train.reshape(60000,28,28,1)
X_test = X_test.reshape(10000,28,28,1)
```

```python
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dense(256, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

**In binary classification we have 2 outputs so we use sigmoid and for multiclassification we have more outputs so we use softmax**

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=10,batch_size=512,)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import gradio as gr
```

```python
def predict_image(img):
  # Gradio provides the image as a numpy array if using the Image input
  img = Image.fromarray(img.astype('uint8')).convert('L')
  img = img.resize((28, 28))
  img_array = np.array(img)
  img_array = img_array.reshape(1, 28, 28, 1)
  img_array = img_array / 255.0
  prediction = model.predict(img_array)
  return {str(i): float(prediction[0][i]) for i in range(10)}
```

```python
gr.Interface(fn=predict_image, inputs=gr.Image(image_mode='L'), outputs="label").launch(debug=True)
```
