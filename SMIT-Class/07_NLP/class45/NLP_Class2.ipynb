**Stemming**    **And**    **Lemitization**

```python
from nltk.stem import PorterStemmer
```

```python
port = PorterStemmer()
```

```python
words = ["playing", "helping", "congratulation","running", "studying", "smiling", "communication"]
```

```python
for word in words:
  print(word+"------->", port.stem(word))
```

**Lemitization**

```python
from nltk.stem import WordNetLemmatizer
```

```python
wordddd = WordNetLemmatizer()
```

```python
words = ["running", "better", "studies", "was"]
```

```python
for word in words:
  print(word+"---------->", wordddd.lemmatize(word))
```

 **Twitter_Sentiment**

```python
import pandas as pd
```

```python
data = pd.read_csv("twitter_training.csv")
```

```python
data.drop(["2401","Borderlands"],axis = 1, inplace=True)
```

```python
data.info()
```

```python
data = data.rename({"im getting on borderlands and i will murder you all ,":  "Text","Positive": "Review"}, axis = 1)
```

```python
data = data[["Text", "Review"]]
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.Review.value_counts().plot(kind = "bar")
```

```python
def all_processing_step(text):
  text   = text.lower()  # lower
  text = re.sub("\s+", " " , text)  # space remove

  text = re.sub("\d+", "" , text)  # decimal remove

  text = "".join([a for a in text if a not in string.punctuation ])  #  Puctuation

  text = " ".join([a for a in text.split() if a not in stopwords.words("english")])  # stopwords

  text = "".join([ wordddd.lemmatize(a) for a in text])  # lemitization

  return text
```

```python
data["Text"]=data["Text"].apply(all_processing_step)
```

```python
data.Review.value_counts()
```

```python
data["Review"]=data.Review.map({
    "Positive":0,
    "Negative":1,
    "Neutral":2,
    "Irrelevant":  3

})
```

```python
import tensorflow
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tok = Tokenizer()
```

```python
tok.fit_on_texts(data["Text"])
```

```python
data["Text"]=tok.texts_to_sequences(data["Text"])
```
