```python
!pip install -q gradio

import os
import subprocess
import gradio as gr

def detect(input_path):
    if input_path is None:
        return None

    results = model_1.predict(source=input_path, save=True)

    save_dir = results[0].save_dir
    filename = os.path.basename(input_path)
    base_name, _ = os.path.splitext(filename)
    output_file_path = None
    for ext in ['.mp4', '.avi', '.mov', '.mkv']:
        temp_path = os.path.join(save_dir, base_name + ext)
        if os.path.exists(temp_path):
            output_file_path = temp_path
            break
    if not output_file_path or not os.path.exists(output_file_path):
        files = [f for f in os.listdir(save_dir) if f.lower().endswith(('.mp4', '.avi', '.mov', '.mkv'))]
        if files:
            output_file_path = os.path.join(save_dir, files[0])
        else:
            output_file_path = os.path.join(save_dir, filename)
    if output_file_path.lower().endswith(('.mp4', '.avi', '.mov', '.mkv')):
        encoded_file_path = os.path.join(save_dir, "browser_compatible_output.mp4")
        command = f'ffmpeg -y -i "{output_file_path}" -vcodec libx264 -pix_fmt yuv420p -acodec aac -f mp4 "{encoded_file_path}"'
        subprocess.run(command, shell=True)
        return encoded_file_path

    return output_file_path

iface = gr.Interface(
    fn=detect,
    inputs=gr.Video(label="Upload Video"),
    outputs=gr.Video(label="Processed Video Result"),
    title="Pothole Detector"
)
iface.launch(debug=True)
```
