```python
# 1. Install EasyOCR for text extraction
!pip install easyocr --quiet

import cv2
import easyocr
from ultralytics import YOLO

# 2. Initialize EasyOCR Reader (English)
reader = easyocr.Reader(['en'], gpu=True)

# 3. Load our trained model
model_1 = YOLO("/content/runs/detect/train/weights/best.pt")

# 4. Run ByteTrack on the video
results = model_1.track(
    source='/content/WhatsApp Video 2026-09-27 at 1.04.51 PM.mp4',
    tracker='bytetrack.yaml',
    persist=True,
    stream=True  # Use stream=True to process frame-by-frame and save RAM
)

print("\n--- Processing Video and Extracting License Plate Text ---\n")

# 5. Process frame by frame, crop plate, and extract text
for frame_idx, result in enumerate(results):
    # Only look at the first 30 frames to demonstrate the text extraction
    if frame_idx >= 30:
        print("\nDemo run completed (Processed first 30 frames).")
        break

    if result.boxes is not None and result.boxes.id is not None:
        # Get original frame image
        frame = result.orig_img

        ids = result.boxes.id.int().tolist()
        boxes = result.boxes.xyxy.int().tolist()

        for track_id, box in zip(ids, boxes):
            x1, y1, x2, y2 = box

            # Crop the license plate from the frame
            cropped_plate = frame[y1:y2, x1:x2]

            if cropped_plate.size > 0:
                # Convert to grayscale for better OCR accuracy
                gray_plate = cv2.cvtColor(cropped_plate, cv2.COLOR_BGR2GRAY)

                # Perform OCR on the cropped plate
                ocr_result = reader.readtext(gray_plate)

                # Combine detected text elements
                plate_text = " ".join([text[1] for text in ocr_result]).strip()

                if plate_text:
                    print(f"Frame {frame_idx} | Plate ID {track_id} | Box: {box} | Extracted Text: '{plate_text}'")
                else:
                    print(f"Frame {frame_idx} | Plate ID {track_id} | Box: {box} | Extracted Text: [No text detected]")
```

```python
import cv2
import easyocr
from ultralytics import YOLO
from collections import defaultdict

# 1. Initialize Reader and Model
reader = easyocr.Reader(['en'], gpu=True)
model_1 = YOLO("/content/runs/detect/train/weights/best.pt")

# 2. Track best crops and high-confidence frames per track ID
best_plate_crops = {}  # track_id -> {'confidence': float, 'crop': ndarray, 'box': list}

# 3. Stream through the video to extract tracks and find the best frame for each plate
results_stream = model_1.track(
    source='/content/WhatsApp Video 2026-09-27 at 1.04.51 PM.mp4',
    tracker='bytetrack.yaml',
    persist=True,
    stream=True
)

print("--- Step 1: Scanning Video and Tracking Best Plate Detections ---")
for frame_idx, result in enumerate(results_stream):
    if result.boxes is not None and result.boxes.id is not None:
        frame = result.orig_img
        ids = result.boxes.id.int().tolist()
        boxes = result.boxes.xyxy.int().tolist()
        confs = result.boxes.conf.tolist()

        for track_id, box, conf in zip(ids, boxes, confs):
            # If this is the highest confidence we've seen for this ID, keep the crop
            if track_id not in best_plate_crops or conf > best_plate_crops[track_id]['confidence']:
                x1, y1, x2, y2 = box
                cropped = frame[y1:y2, x1:x2]
                if cropped.size > 0:
                    best_plate_crops[track_id] = {
                        'confidence': conf,
                        'crop': cropped,
                        'box': box,
                        'frame_idx': frame_idx
                    }

print(f"Scanning complete! Found {len(best_plate_crops)} unique plate IDs.")
```

```python
print("--- Step 2: Running OCR Only on the Best Image of Each Plate ---")

for track_id, info in best_plate_crops.items():
    crop = info['crop']
    gray_plate = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)

    # Run EasyOCR only once per unique plate track
    ocr_result = reader.readtext(gray_plate)
    plate_text = " ".join([text[1] for text in ocr_result]).strip()

    print(f"Plate ID {track_id} | Best Frame: {info['frame_idx']} | Conf: {info['confidence']:.2f} | Extracted Text: '{plate_text or '[No Text Detected]'}'")
```
