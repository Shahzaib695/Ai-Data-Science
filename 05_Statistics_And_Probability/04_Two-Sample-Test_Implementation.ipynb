```python
import numpy as np
from scipy import stats
```

```python
group_A = [85,88,90,92,87,85,89,91,86,88]
group_B = [82,84,80,93,81,79,78,85,84,83]
```

```python
print('Null Hypothesis of this is that there is no difference between them')
print('Alternate Hypothesis of this is that there is a difference between them')
```

```python
t_stats,p_values = stats.ttest_ind(group_A,group_B,equal_var=False)
```

```python
t_stats
```

```python
alpha = 0.05
```

```python
if p_values < alpha:
    print("Null Hypothesis rejected")
else:
    print("Null HYpothesis Accepted")
```
