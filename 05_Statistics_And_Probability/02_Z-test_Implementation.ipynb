```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172,174,168,169,171,173,175,170,169,172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n=len(sample)
```

```python
z_score = (sample_mean - population_mean)/ (population_std/np.sqrt(n))
z_score
```

```python
p_value = 2 * (1-norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05
# in p-values when the calculated p-value is less than alpha we reject our null hypothesis and if it is greater than alpha then we accept it our null hypothesis
if p_value < alpha:
    print("We will reject the null hypothesis")
else:
    print("We accept the null hypothesis")
```
