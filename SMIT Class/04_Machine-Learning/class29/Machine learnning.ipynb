```python
import pandas as pd
```

```python
data = pd.read_csv("weight-height.csv")
```

```python
data.head()
```

## X,y

```python
x = data[["Height"]]
y = data[["Weight"]]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
X_train
```

### Model traning

```python
from sklearn.linear_model import LinearRegression
```

```python
model  = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_train)
```

### Training data

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
r2_score(y_train, y_pred)*100   # accuracy
```

```python
mean_absolute_error(y_train, y_pred)  # lose
```

### Testing Data

```python
y_test_pred = model.predict(X_test)  
```

```python
r2_score(y_test, y_test_pred)*100  # accuracy
```

```python
mean_absolute_error(y_test, y_test_pred)  # lose
```

### overfitting:
            Differnec (Gap) between Trainig and Testing Acccuracy

### underfit: performance low on training data    perfromace low on testing data
### overfit:  performance high on traning data    perfromance low on testing data
### Bestfit:  performancce high on training data  performance high on testing data

```python
def wight_prediction(height):
    return model.predict([[height]])*0.453592
```
