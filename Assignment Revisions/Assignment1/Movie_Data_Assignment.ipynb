### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.sort_values(by="Rating_100_Scale", ascending=False).head(5)
top5
```

```python
x = np.arange(5)
width = 0.35

plt.figure(figsize=(10,6))
plt.bar(x - width/2, top5['Meta_score'].values, width, label='Meta_score')
plt.bar(x + width/2, top5['Rating_100_Scale'].values, width, label='IMDB_Rating (scaled)')

plt.xticks(x, top5['Series_Title'].values, rotation=45, ha='right')
plt.ylabel('Score (out of 100)')
plt.title('Top 5 Movies: Meta_score vs IMDB Rating (scaled)')
plt.legend()
plt.tight_layout()
plt.show()
```
