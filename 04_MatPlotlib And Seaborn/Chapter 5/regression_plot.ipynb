## Regression Plot
    Regression plot is not that important we only use it for machine learning algorithm purpose we will not focus on when and where it is used we will see only how it is created

    To create it we first create x and y axis we usually do prediction in machine learning the thing we want to predict goes in y-axis and through which we are going to predict goes in x-axis. To create these plot we create a scatter plot in this plot after doing some calculations a best fit line is drawn. This is how regression plot is created inshort.
    

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.lmplot(x='total_bill',y='tip',data = tips)
```



In the attached chart the line on x axis suggest that there is a bill approx 45 dollars what could be the tip of it we are using just for an example purpose like the tip we will recieve would be approximately 5.6 dollars something like that i hope u got my point. This is the way we create linear regression algorithms

```python
# lets see can we create hue (difference)
sns.lmplot(x='total_bill',y='tip',data = tips,hue='sex')
```

    Here in the above plot we can see different lines for both genders
