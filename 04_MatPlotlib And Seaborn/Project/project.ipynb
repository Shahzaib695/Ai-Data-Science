### Loading the Libraries and Dataset

1. Which team won the most matches?

```python
df['match_winner'].value_counts()
```

```python
plt.title('Most Match Win By Team')
match_wins = df['match_winner'].value_counts()
sns.barplot(y=match_wins.index,x=match_wins.values,palette='rainbow')
# to fix overlapping i am switching their positions
```

2. Toss Decision Trends

```python
df.head(2)
```

```python
sns.countplot(x=df['toss_decision'])
plt.title("Toss Decision Trends ")
```

3. Toss Winner vs Match Winner

```python
df.head(1)
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count*100)/df.shape[0]
# there are 36 teams who won the toss along with the match
percentage.round(2)
```

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
df.head(1)
```

```python
key_player = df['player_of_the_match'].value_counts().head(10)
```

```python
key_player
```

```python
plt.title('Top 10 Players With Most MOM Awards')
sns.barplot(y=key_player.index,x=key_player.values,palette='mako')
```

2 Top Scorers

```python
df.head(1)
```

```python
high_scorers = df.groupby(df.top_scorer)['highscore'].sum().sort_values(ascending=False).head(2)
```

```python
high_scorers.plot(kind='barh')
```

10 Best Bowling Figures

```python
df.head(1)
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
```

```python
df['highest_wickets'] = df.highest_wickets.astype(int)
```

```python
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
top_bowlers
```

```python
top_bowlers.plot(kind = 'barh')
```
