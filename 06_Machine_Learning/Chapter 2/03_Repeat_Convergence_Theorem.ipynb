## **Repeat Convergence Theorem**

    Lets create a gradient decent for simple linear regression it is drawn something like this (attaching the image below) You can see it is something like a curve. For better understanding see the image in notebook or refer to notes in the notes folder.



    For suppose we calculated J of theta 1 and we got a point so the system from which linear regression determines that the point will go down to form a curve is what we call as "Repeat Convergence Theorem".

    There is another thing in this theorem which is Learning Rate. We keep learning rate low because the value of loss function can varry a bit too much if we keep the learning value high so keeping that value low enables the global minima and the loss value consistent.
