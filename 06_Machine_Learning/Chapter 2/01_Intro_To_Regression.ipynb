# **Difference Between Regression And Classification Algorithm**

    We use regression algorithm when we need to predict something which is continous like salary of employees working in a company. While for classification algorithm suppose u r doing predictions on titanic dataset where u have to classify whether this person survived or not. In short we use classification algorithm when we need to classify something and regression algorithm when we need to predict a continous data

## **What is basically Regression?**

    Regression is a way to predict a number. It helps to find out how one thing changes when something else changes.
    Regression is about finding patterns in numbers so we can make smart predictions.
