## **Overfitting**
    We say a model is overfit if it performs better on training dataset while on testing dataset it perfroms bad.
    When a model performs better on training data we say it as low bias.
    When a model doesnt performs better on testing data we say it as high variance.
## **Underfitting**
    We say a model is underfit when it doesn't performs better on training dataset and even on testing dataset it perfromance remains worse.
    When a model doesn't performs better on training data we say it as high bias.
    When a model doesnt performs better on testing data we say it as high variance.
