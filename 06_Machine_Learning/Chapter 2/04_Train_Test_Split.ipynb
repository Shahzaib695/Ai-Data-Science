## Explaination for training and testing

    When we create a model for a particular dataset we split our dataset in training and testing for example we are using 80% of our data for training of our model and leaving 20% for testing of our model like how much its accurate or not this the whole case scenario. Then we compare the predicted data and the actual data to check the difference between them using various things like mean squared error, r2 score and much more things we will see properly in our project.
