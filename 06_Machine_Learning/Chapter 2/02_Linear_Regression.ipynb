## **SLOPE**:
    The slope is the measure of steepness and direction of line. It tells us how much the output changers for every single unit increase in input. For example if we have a slope of 1 it means for every 1 step forward on x-axis we will move 1 step up on y-axis.

    By changing the slope value we tilt the line on our graph which helps the model to find the best fit for our data.
