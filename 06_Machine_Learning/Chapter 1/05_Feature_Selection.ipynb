# **Feature Selection**
Feature Selection means selecting the most useful features and removing the rest.

**Why is it important?**

    Answer is simple it is important because it helps in reducing noise and overfitting in data, speeds up training,improving model accuracy, make model interpretation easier.


### **In machine learning we do feature selection based on most probably two ways**
#### **1. Filter Methods (Pure Statistics):**

**Correlation Matrix:**

        Removes highly correlated features.

**Chi-Square Test (Categorical vs Categorical)**


**Annova F-test (Numerical vs Categorical target)**

#### **2. Embedded Methods (Selection built into the model):**
**Lasso Regression:**

        Shrinks coefficent to 0

**Tree Based Models (Random Forest, XGBoost):**

        Feature Important Scores.
