# Types Of Machine Learning

**Machine learning has 3 major types:**


#### **1- Supervised Learning:**
Supervised learning is like a student being taught by a teacher. We give the machine input and the correct answer and it learns to predict.
    Lets see with an example:



Mostly machine learning applications are based on supervised learning. When we give data to models we are basically giving two types of variable input variables and output variables. Like in the above example we can cleary see this a data which is used for creating a pattern to predict that a person will recieve loan or not. The income and credit score are the key input variable which will be used for determining that a person will recieve the loan or not. When we will give the data to machine it will learn from this data and it will create a pattern on its own and will use this pattern for predicting future data. The loan column is an output variable.

#### **2- Un Supervised Learning:**
In Unsupervised learning we will give the data to machine but we will not predict anything we are just going to find patterns, group similar items, reduce complexity, spot outliers.
The best example of this is google search for example we search for python on google and our resutls will only show info related to python. Unsupervised learning is used for training supervised learning models sounds strange but we will see in upcoming chapters more properly.

#### **3- Reinforcement Learning:**
Reinforcement learning is like training a dog. You dont teach it directly instead you reward good behavior and ignore or punish bad habbits. Overtime it learns what actions gives it threats and what action gives it treats. Thats the core idea of Reinforcement learning--learning by trial and error.
