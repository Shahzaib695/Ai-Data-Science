# Data Preprocessing
**Data preprocessing refers to prepare clean data so it can be analyzed or used in a machine learning model.**

If data cleaning means to fix mistakes. Then data preprocessing is about transforming valid data into a usable format.

## Steps Of Data Preprocessing

#### **1. Encoding Text Variables:**

    The first step is to convert our text labels (like male,yes,southeast) into numbers because most of our machine learning algorithms dont understand string variables so its neccessary to convert them into numbers. For example u have a gender column u can encode it into numbers like 0 for male and 1 for female this would help machine to understand the data.
##### **Types Of Encoding:**
**There are 2 types of encoding in a data:**
**1. Label Encoding:**

    It is done for ordered categories like gender. We have discussed that example in the above examples.

**2 . One Hot Encoding:**

    Its for non ordered categories like region for example in a data u have a column named as regions and in that columns there are multiple regions like north,south,east,west we can swap them with numbers but its not the ideal situation we can do a much more better way its like we create new column for regions. New columns name could be region_north and what it will do is use 1 if the resident reside in north region and 0 if the resident dont reside in north region it will do the same for every region this is one hot encoding.

#### **2. Feature Transformation (log,Square root,etc):**
    It is used for handling skewed data like right skewed or left skewed data we will properly see in the project. If u are wondering what skewed is refer to stats folder and see the formatted pdf or original odt whatever u wish its up to u :).

#### **3. Feature Scaling (Normalization or Standardization):**
    Bring numerical values to the same scale it is especially useful for decision based algorithms. To understand this lets look at an example suppose u have two columns named as age and salary. U plot a linear regression scatter plot u noticed that the line of liner regression is not proper its messed up as they are not on same scale. To fix that we can scale all the values from 0-1 like for ages column if the age is 26 we can convert it to 0.26 and for salary suppose the salary of someone is 36,000 we can convert it to 0.36. Now the data from  both columns lies under 0-1 this means we can properly scale them and linear regression plot can be created in a more proper way i hope u got my point.
