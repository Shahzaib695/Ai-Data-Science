# **Feature Engineering**
**Feature Engineering means creating new features or transforming existing ones to expose useful patterns that Machine Learning models can learn from**

A Question might come in your mind why do we need it?

    The answer is simple ML models dont have domain logic we have to give them the right signals. 

## **Feature Engineering Techniques In Machine Learning**
**The common techniques used in machine learning are as follows:**


**1. Mathematical Combinations**

**2. Target Based Flags**

**3. Binning (when it helps)**

**4. Time based Features (if it exists)**
