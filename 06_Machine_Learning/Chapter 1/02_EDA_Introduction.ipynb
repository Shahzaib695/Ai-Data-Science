# **EDA** 
EDA stands for Exploratory Data Analysis it is the step where we explore the data too:

    Understand it
    Discover patterns
    Spot anomalies (outliers, nan values)
    Generate Insights
    And decide what to do next
EDA is a crucial thing that needs to be done before building a machine learning model.    

## **STEPS OF EDA**
#### **1. Viewing The Data:**
    View the data using head(),shape(),info().
    What columns do we have? What type of data we have?
#### **2. See Statistical Summary Of Data:**
    See mean,median,mode,std,max,min,quartiles we can see all these together using describe().
    Helps understand spread and centeral tendency.
#### **3. See Value Counts:**
    See how many unique value are in a specific column.
    Its great for categorical columns.
#### **4. Missing Value Analysis:**
    Analyze crucial things like Where are the gaps? What percent of data is missing?
#### **5. Visualizations:**
    Visualize data for properly understanding it. We mostly create these types of plots.
**Histogram** -> **To Visualize Distribution Of Values**

**BoxPlot** -> **To Visualize Outliers and Spread**

**BarPlot** -> **To Visualize Comparision Of Categories**

**Correlation Heatmaps** -> **To Visualize Linear Relationships Between Numerical Features**

**Scatter Plot** -> **To Visualize Bivariate Relationships**

#### **6. Target Variable Exploration:**
    How does your output (like 'charges' in your dataset) relate to your variable. In simple words we see how our input variable relates to our output variable on which we will be making some predictions.



# Data Cleaning
    After doing EDA we do the most important thing which is to clean the data.
## Steps To Clean Data Are As Follows:
#### **1. Handling Missing Values:**
    Check which columns have missing values (nulls).
**Strategies To Handle:**
    Drop missing rows/columns (only if very few).
    
**Impute With:**
    Mean/Median -> For numerical data.
    Mode -> For categorical data.
    Advanced -> Linear Regression, KNN, or interpolation (for future learning).
